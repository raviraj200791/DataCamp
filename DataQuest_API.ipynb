## World Bank's Development Indicators API

```python
url = "https://api-server.dataquest.io/economic_data/countries"
response = requests.get(url)
```

```python
data = json.loads(response.json())
```

```python
print("Total countries: {cnt}".format(cnt=len(data)))
print("First country: {cntry}".format(cntry=data[0]["short_name"]))
```

```python
params = {
    "filter_by": "region=Sub-Saharan Africa"
}
response = requests.get(url, params = params)
data = json.loads(response.json())
print("Total countries: {cnt}".format(cnt=len(data)))
print("First country: {cntry}".format(cntry=data[0]["short_name"]))
```

```python
params = {
    "filter_by": "region=Sub-Saharan Africa",
    "limit" : 5
}
response = requests.get(url, params = params)
data = json.loads(response.json())
print("Total countries: {cnt}".format(cnt=len(data)))
print("First country: {cntry}".format(cntry=data[0]["short_name"]))
```

## Sending Data with POST Requests

```python
url = "https://jsonplaceholder.typicode.com/posts"
new_post = {
    "title": "Learning Python API",
    "body": "POST requests lets you send data to APIs",
    "userId": 101
}

response = requests.post(url, json = new_post)

if response.status_code ==201:
    print("Post created successfully.")
else:
    print("Someting went wrong: {code}".format(code = response.status_code))
```

## Error Handling and Best Practices for Python APIs

```python
try:
    response = requests.get("http://api.open-notify.org/iss-now.json", timeout = 10)
    response.raise_for_status()
    data = response.json()
    print(f"Current ISS position: {data['iss_position']}")
except requests.exceptions.Timeout:
    print("The ISS tracking API isn't responding. Try again in a moment.")
```

## Managing Rate Limits

```python
import time
```
